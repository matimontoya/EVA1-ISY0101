2. Importar librerías

```python
import os

from google.colab import userdata

from langchain_google_genai import ChatGoogleGenerativeAI

from langchain_core.documents import Document

from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_community.vectorstores import FAISS

print("Librerías importadas correctamente")
```

3. Configurar Gemini

```python
API_KEY = userdata.get("GOOGLE_API_KEY")

if not API_KEY:
    raise ValueError("No se encontró GOOGLE_API_KEY en los secretos de Colab")

os.environ["GOOGLE_API_KEY"] = API_KEY

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

print("Gemini 3.5 Flash Lite conectado correctamente")
```

4. Cargar los documentos PDF

```python


from google.colab import files

archivos_subidos = files.upload()

print("\nArchivos cargados:")

for nombre in archivos_subidos:
    print("-", nombre)
```

5.
Extraer texto de los PDF

```python
from pypdf import PdfReader

documentos = []

for nombre in archivos_subidos:

    reader = PdfReader(nombre)

    texto = ""

    for pagina in reader.pages:

        contenido = pagina.extract_text()

        if contenido:
            texto += contenido + "\n"

    documentos.append({
        "fuente": nombre,
        "texto": texto
    })

    print("=" * 70)
    print(f"Archivo: {nombre}")
    print(f"Páginas: {len(reader.pages)}")
    print(f"Caracteres extraídos: {len(texto):,}")
    print("=" * 70)
```

6. Revisar el texto extraído

```python
for documento in documentos:

    print("=" * 80)
    print(f"FUENTE: {documento['fuente']}")
    print("=" * 80)

    print(documento["texto"][:1500])

    print("\n")
```

7. Crear chunks

```python
import re

articulos = []

for documento in documentos:

    texto = documento["texto"]

    patron = r"(?=(?:ARTICULO|ARTÍCULO|Artículo)\s+[A-ZÁÉÍÓÚÑ0-9]+)"

    partes = re.split(patron, texto)

    for parte in partes:

        parte = parte.strip()

        if len(parte) < 100:
            continue

        if re.match(
            r"(?:ARTICULO|ARTÍCULO|Artículo)\s+",
            parte
        ):
            articulos.append(
                Document(
                    page_content=parte,
                    metadata={
                        "fuente": documento["fuente"]
                    }
                )
            )


text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1500,
    chunk_overlap=250
)

chunks = []

for articulo in articulos:

    textos = text_splitter.split_text(
        articulo.page_content
    )

    for texto in textos:

        chunks.append(
            Document(
                page_content=texto,
                metadata={
                    "fuente": articulo.metadata["fuente"],
                    "articulo": articulo.page_content.split(":")[0].strip()
                }
            )
        )

print(f"Artículos detectados: {len(articulos)}")
print(f"Total de chunks creados: {len(chunks)}")
```

8. Revisar chunks

```python
for i, chunk in enumerate(chunks[:3]):

    print("=" * 80)
    print(f"CHUNK #{i}")
    print(f"Artículo: {chunk.metadata['articulo']}")
    print(f"Fuente: {chunk.metadata['fuente']}")
    print("=" * 80)
    print(chunk.page_content)
    print()
```

9. Crear embeddings

```python
from sentence_transformers import SentenceTransformer
from langchain_core.embeddings import Embeddings


class LocalEmbeddings(Embeddings):

    def __init__(self):

        self.model = SentenceTransformer(
            "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
        )

    def embed_documents(self, texts):

        return self.model.encode(
            texts,
            normalize_embeddings=True
        ).tolist()

    def embed_query(self, text):

        return self.model.encode(
            [text],
            normalize_embeddings=True
        )[0].tolist()


embeddings = LocalEmbeddings()

print("Modelo de embeddings cargado correctamente")
```

10. Crear índice FAISS

```python
vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)

print("Índice FAISS creado correctamente")
print(f"Cantidad de vectores: {vectorstore.index.ntotal}")
```

11. Query Rewriting

```python
pregunta = "¿Qué procedimiento judicial se aplica cuando alguien deja de pagar durante 6 meses?"

prompt_reescritura = f"""
Eres un asistente especializado en recuperación de información
dentro de un reglamento de copropiedad inmobiliaria.

Transforma la pregunta del usuario en una consulta breve
para búsqueda semántica dentro del reglamento.

Reglas estrictas:
- Mantén exactamente la intención de la pregunta.
- No respondas la pregunta.
- Usa como máximo 6 conceptos o términos.
- Prioriza palabras que probablemente aparezcan literalmente
  en el reglamento.
- No agregues consecuencias, sanciones, procedimientos ni
  situaciones que el usuario no haya mencionado.
- No agregues conceptos como abandono, ausencia, morosidad,
  cobranza o suspensión de servicios, salvo que estén
  explícitamente presentes en la pregunta.
- No cambies el significado de la pregunta.
- Devuelve solamente la consulta de búsqueda.

Pregunta del usuario:
{pregunta}
"""

respuesta_reescritura = llm.invoke(
    prompt_reescritura
)

if isinstance(respuesta_reescritura.content, list):

    consulta_mejorada = (
        respuesta_reescritura.content[0]["text"].strip()
    )

else:

    consulta_mejorada = (
        respuesta_reescritura.content.strip()
    )

print("PREGUNTA ORIGINAL:")
print(pregunta)

print("\nCONSULTA OPTIMIZADA:")
print(consulta_mejorada)
```

12. Recuperación con MMR

```python
retriever_mmr = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 4,
        "fetch_k": 10,
        "lambda_mult": 0.7
    }
)

resultados_mmr = retriever_mmr.invoke(
    consulta_mejorada
)

print("PREGUNTA ORIGINAL:")
print(pregunta)

print("\nCONSULTA OPTIMIZADA:")
print(consulta_mejorada)

print("\n" + "=" * 80)
print("RESULTADOS MMR")
print("=" * 80)

for i, documento in enumerate(resultados_mmr, 1):

    print(f"\n--- RESULTADO {i} ---")
    print(f"Fuente: {documento.metadata['fuente']}")
    print(documento.page_content[:800])
    print("-" * 80)
```

13. Construir contexto

```python
contexto_final = "\n\n".join(
    [
        f"Artículo: {documento.metadata['articulo']}\n"
        f"Fuente: {documento.metadata['fuente']}\n"
        f"{documento.page_content}"
        for documento in resultados_mmr
    ]
)

print("Contexto construido correctamente.\n")
print(contexto_final)
```

14. Generar respuesta final

```python
prompt_respuesta = f"""
Eres un asistente especializado en el Reglamento de Copropiedad
Inmobiliaria de Chile.

Responde la pregunta del usuario utilizando únicamente la información
contenida en el contexto recuperado.

Reglas estrictas:
- Responde directamente la pregunta.
- No inventes información.
- No utilices conocimientos externos al contexto.
- Explica la respuesta de forma clara y sencilla.
- Identifica el artículo que fundamenta la respuesta.
- Solo menciona un artículo si su número aparece explícitamente
  en el contexto.
- No deduzcas ni calcules el número del artículo.
- Si el artículo no puede determinarse con seguridad, indícalo.
- Menciona la fuente del documento utilizado.
- No menciones "Documento 1", "Documento 2", etc.
- No describas el funcionamiento del RAG.
- No menciones que eres una IA.

Formato:

Respuesta:
[respuesta directa y clara]

Fundamento:
[artículo que respalda la respuesta]

Fuente:
[nombre del documento utilizado]

Pregunta del usuario:
{pregunta}

Contexto recuperado:
{contexto_final}
"""

respuesta_final = llm.invoke(
    prompt_respuesta
)

if isinstance(respuesta_final.content, list):

    respuesta = (
        respuesta_final.content[0]["text"].strip()
    )

else:

    respuesta = (
        respuesta_final.content.strip()
    )

print("PREGUNTA:")
print(pregunta)

print("\n" + "=" * 80)
print("RESPUESTA FINAL DEL RAG")
print("=" * 80)

print(respuesta)
```
